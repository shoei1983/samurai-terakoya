```python
"""if文を用いて、以下の条件分岐をするコードを記述してください。値はprint関数で出力してください。

・変数varが、3の倍数の場合は「Fizz」を出力
・変数varが、5の倍数の場合は「Buzz」を出力
・変数varが、3の倍数と5の倍数の両方に該当する場合は「FizzBuzz」を出力
・上記のどの場合にも該当しない場合は、変数varの値を出力

ただし、変数varは正の整数とします。"""


#ランダムな整数を利用するために、randomモジュールをインポート
import random

# 変数varに0～4までのランダムな整数を代入する
var = random.randint(1, 15)

# 変数varの値を出力
print(var)

""" ・変数varが、3の倍数と5の倍数の両方に該当する場合は「FizzBuzz」を出力
 1.if varを15でで割った余りがゼロとなるとき
 2.elif varを5で割った余りがゼロとなるとき　Buzz
 3.elif varを3で割った余りがゼロとなるとき　Fizz
 4.else どれにも該当しないとき　変数varの値を出力"""

if var % 15 == 0:
    print("FizzBuzz")
elif var % 5 == 0:
    print("Buzz")
elif var % 3 == 0:
    print("Fizz")
else:
    print(var)




```
