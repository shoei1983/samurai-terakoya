```python
# 11章の次の課題

# forループで水～冥の9行を表示し、つぎにwhileループで水～冥の9行を表示してください

array = ["水","金","地","火","木","土","天","海","冥"]

# forいこうか
for star in array:
  print(star)


# whileいこうか カウント変数iが8になるまで繰り返しかな キーは8がMAX
# まずはiの定義

i = 0

while i < 9:
  # キー　i番目を表示
      print(array[i])
  # iに1ずつ加算　カウンターってやつ
      i = i + 1
```
