```python
# 15章　課題
"""
名前(name)と年齢(age)の属性を持つHumanクラスを作成してください。
nameとageを標準出力(print)するメソッド(printinfo)を追加してください。

Humanクラスのインスタンスは、変数に代入してプログラム内で使用してください
"""

# クラス定義
class Human:
    # コンストラクタを定義する!属性設定！
    def __init__(self, name, age):# ←教えて！ __init__ って何者？selfって何者？
      self.name = name# ←教えて！self.nameとnameの違いは？
      self.age = age# ←教えて！self.ageとageの違いは？

    # メソッド定義！名前と年齢を出力するやつ！
    def printinfo(self):# ←教えて！selfの意味は？
      print(self.name, self.age)# ←教えて！self.nameってnameだけじゃだめなの？


# インスタンス化！
human = Human("侍田太郎", 36)# ←教えて！侍田太郎がself.nameってこと？

# 属性にアクセスし、値を出力する
human.printinfo()# ←教えて！()の中が空欄なのなんかキモイw


"""
AIによる質問への回答を記載

initって何者？：init初期化儀式クラス誕生時に必須。
selfって何者？：selfはインスタント(私)自身。

self.nameとnameの違いは？：self.nameは箱。nameは中身。
self.nameってnameだけじゃだめなの？：
　インスタントが複数ある場合をイメージすると理解しやすい。
　どのインンスタンスかわからなくなる。

()の中が空欄なのなんかキモイw：引数が不要だから



"""
```
